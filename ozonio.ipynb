```python
#Importar bibliotecas

import pandas as pd
```

```python
import numpy as np
```

```python
#Leitura dos arquivos

df = pd.read_csv("2025.csv", sep=';', skiprows=1)
display(df.head())
```

```python
#Remover colunas QA/QC

df = df.loc[:, ~df.columns.str.contains('QA/QC', case=False)]
display(df.columns)
```

```python
#Renomear coluna Data e Hora

df.rename(columns={'Data e Hora': 'DataHora'}, inplace=True)
display(df.head())
```

```python
#Conversão das datas

meses = {'jan':'01', 'fev':'02', 'mar':'03', 'abr':'04', 'mai':'05', 'jun':'06', 'jul':'07', 'ago':'08', 'set':'09', 'out':'10', 'nov':'11', 'dez':'12'
} 

for pt, num in meses.items():
    df['DataHora'] = df['DataHora'].str.replace(pt, num)


df['DataHora'] = pd.to_datetime(
    df['DataHora'],
    format='%d-%m-%Y %H:%M'
)

print(df['DataHora'].dtype)
```

```python
#Transformar DataHora em índice

df = df.set_index('DataHora')
```

```python
#Relatório anual de 2025 (21, 22, 23, 24 e 25) ano bissexto: 366 dias x 24 horas = 8784 , cada coluna deveria ter 8784 valores

relatorio = pd.DataFrame()

relatorio['Dados Válidos'] = df.count()

relatorio['NaN'] = df.isna().sum()

relatorio['Cobertura (%)'] = (relatorio['Dados Válidos'] / len(df)) * 100

relatorio['Falha (%)'] = (relatorio['NaN'] / len(df)) * 100 

display(relatorio)
```

```python
#Filtrar série boa anual

series_validas = relatorio[relatorio['Cobertura (%)'] >= 75]
display(series_validas)
```

```python
#Lista das colunas válidas

colunas_validas = series_validas.index.tolist()
display(colunas_validas)
```

```python
#Criação da base validada

df_ano_validado = df[colunas_validas]

df_ano_validado = df_ano_validado.replace(',', '.', regex=True)

df_ano_validado = df_ano_validado.apply(pd.to_numeric, errors='coerce')
```

```python
df_ano_validado['Data'] = (df_ano_validado.index.date)
```

```python
#Horas válidas por dia

horas_validas = (df_ano_validado.groupby('Data').count())
#display(horas_validas)
```

```python
#Gap de horas indisponíveis

for coluna in df_ano_validado.columns:
    serie_nan = df_ano_validado[coluna].isna()

    grupos = (serie_nan != serie_nan.shift()
    ).cumsum()

    tamanhos = serie_nan.groupby(grupos).sum()

    maior_gap = tamanhos.max()

    print(f'{coluna}: maior gap = {maior_gap} horas')
```

```python
colunas_o3 = [col for col in df_ano_validado.columns if 'Ozônio' in col]

display(colunas_o3)
```

```python
df_o3 = df_ano_validado[colunas_o3].copy()

display(df_o3.head())
```

```python
media_movel_8h = (df_o3.rolling(window=8, min_periods=6).mean())

display(media_movel_8h.head(15))
```

```python
max_diaria_o3 = (media_movel_8h.resample('D').max())

display(max_diaria_o3.head())
```

```python
ultrapassagens_o3 = pd.DataFrame(index=max_diaria_o3.columns)

ultrapassagens_o3['PI-1 (140)'] = (max_diaria_o3 >= 140).sum()

ultrapassagens_o3['PI-2 (130)'] = ((max_diaria_o3 >= 130) & (max_diaria_o3 < 140)).sum()

ultrapassagens_o3['PI-3 (120)'] = ((max_diaria_o3 >= 120) & (max_diaria_o3 < 130)).sum()

ultrapassagens_o3['PI-4 (110)'] = ((max_diaria_o3 >= 110) & (max_diaria_o3 < 120)).sum()

ultrapassagens_o3['PF (100)'] = ((max_diaria_o3 >= 100) & (max_diaria_o3 < 110)).sum()

display(ultrapassagens_o3)
```

```python
media_mensal_o3 = (max_diaria_o3.resample('ME').mean())

display(media_mensal_o3)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

for coluna in media_mensal_o3.columns:
    plt.plot(media_mensal_o3.index, media_mensal_o3[coluna], label=coluna)

plt.title('Médias Mensais de Ozônio (O3) - 2025')

plt.xlabel('Meses')

plt.ylabel('Concentração de O3 [µg/m³])')

plt.grid(True)

plt.legend()

plt.tight_layout()

plt.show()
```
